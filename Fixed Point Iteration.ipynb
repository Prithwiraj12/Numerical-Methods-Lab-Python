```python
def f(x):
  return x**2-3*x+1
def g(x):
  return (x**2+1)/3
def gd(x):
  return (2*x)/3
```

```python
#Initial Setup
xi = float(input())
n = int(input())
es = 0.5*10**(2-n)
ea = 100
it = 1
```

```python
def sf(num):
  return float(f"{num:.{n}g}")
```

```python
while ea >= es:
  if(abs(gd(xi)) > 1):
    print("Divergence")
    break
  else:
    xi1 = sf(g(xi))
    ea = abs((xi1-xi)/xi1)*100
    print("Iteration No: ", it, "Root: ", xi1)
    xi = xi1
    it = it+1
```
