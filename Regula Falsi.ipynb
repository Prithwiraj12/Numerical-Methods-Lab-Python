```python
def f(x):
  return x**2-2
```

```python
#Initial Setup
xl = int(input("Enter Lower Limit XL: "))
xu = int(input("Enter Upper Limit XU: "))
n = int(input("Enter Significant Figure n: "))
es = 0.5*10**(2-n)
ea = 100
it = 1
```

```python
def sf(num):
  return float(f"{num:.{n}g}")
```

```python
while ea >= es:
  xr = xu - ((sf(f(xu)) * (xl - xu)) / (sf(f(xl)) - sf(f(xu))))
  xr = sf(xr)
  if f(xr)*f(xu) < 0:
    xl = xr
  elif f(xl)*f(xr) < 0:
    xu = xr

  cv = xr
  if it == 1:
    pv = cv
  elif it >= 2:
    ea = abs((cv-pv)/cv)*100
    pv = cv

  print("Iteration No: ", it, "->", cv)
  it = it+1
```
