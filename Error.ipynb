```python
import math
```

```python
x = float(input("Enter x: "))
```

```python
es = float(input("Enter Stopping Creterion: "))
```

```python
TV = math.exp(x)
TV
```

```python
ca = 0
pa = 0
n = 0
```

```python
print(f"\n{'n':>2} {'Term':>12} {'Approximation':>15} {'Ea':>10}{'Es(%)':>10}{'Et':>10}")
print("-"*65)

while True:
  term = (x**n)/math.factorial(n)
  ca = ca+term

  ea = (abs(ca-pa))/ca
  et = (abs(TV-ca))/TV

  print(f"{n:2d}{term:12.6f}{ca:15.6f}{ea:10.6f}{es:10.4f}{et:10.6f}")
  if ea < es:
    break
  pa = ca
  n+=1;



```
