```python
#Modified Euler
```

```python
def f(x, y):
  return x+y
```

```python
#initial setup
h = 0.05
x = 0
y = 1
xf = 0.1
```

```python
while x < xf:
  yp = y + f(x, y)*h
  ycp = 100
  ycprev = -100
  while ycprev != ycp:
    ycprev = ycp
    ycp = y + ((f(x, y) + f(x+h, yp))/2) * h
    yp = ycp
  x = x+h
  y = yp
  print(x, y)



```
