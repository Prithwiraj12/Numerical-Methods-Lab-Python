```python
import numpy as np
```

```python
n = int(input("Ener Marix Size: "))
A = np.zeros((n, n))
for i in range(n):
  for j in range(n):
    A[i][j] = float(input(f"A[{i}][{j}] = "))
```

```python
print(A)
```

**"Direct Library"**

```python
if np.linalg.det(A) != 0:
  print("1-Norm = ", np.linalg.cond(A, 1))
  print("2-Norm = ", np.linalg.cond(A, 2))
  print("Infinity-Norm", np.linalg.cond(A, np.inf))
else:
  print("Matrix is Singular")
```

**Manual Calculation**

```python
A_inv = np.linalg.inv(A)

col_sum1 = []
col_sum2 = []

for j in range(n):
  sum1 = 0
  sum2 = 0
  for i in range(n):
    sum1 = sum1 + abs(A[i][j])
    sum2 = sum2 + abs(A_inv[i][j])
  col_sum1.append(sum1)
  col_sum2.append(sum2)

condA = max(col_sum1) * max(col_sum2)
condA
```

```python
A_inv = np.linalg.inv(A)

row_sum1 = []
row_sum2 = []

for i in range(n):
  sum1 = 0
  sum2 = 0
  for j in range(n):
    sum1 = sum1 + abs(A[i][j])
    sum2 = sum2 + abs(A_inv[i][j])
  row_sum1.append(sum1)
  row_sum2.append(sum2)

condA = max(row_sum1) * max(row_sum2)
condA
```
