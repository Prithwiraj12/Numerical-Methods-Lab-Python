```python
## Cramers' rule
n = int(input("Number of Equation "))
```

```python
#Coefficient Matrix Input
A = []
print("Coefficient Matrix:")
for i in range(n):
  row = list(map(float, input().split()))
  A.append(row)
A = np.array(A)
```

```python
B = []
for i in range(n):
  B.append(float(input()))
B = np.array(B)
```

```python
D = np.linalg.det(A)
D
```

```python
if D == 0:
  print("No Unique Solution exists")
else:
  for i in range(n):
    Ai = A.copy()
    Ai[:,i] = B
    Di = np.linalg.det(Ai)
    x = Di/D
    print(round(x))
    print(x)
```
