```python
def f(x):
  return 0.2+25*x-200*(x**2)+675*(x**3)-900*(x**4)+400*(x**5)
```

```python
#Simpson 1/3 Rule
n = 2
x0 = float(input("Lower Limit: "))
x2 = float(input("Upper Limit: "))
h = (x2-x0)/n
x1 = x0+h
```

```python
I = (1/3)* h * (f(x0)+4*f(x1)+f(x2))
print("Area :", I)
```

```python
#Multiple Segment Simpson 1/3 rule

n = int(input("Enter the number of Segment: "))
if n%2 == 1:
  print("Simpson rule can not be applied")
else:
  x0 = float(input("Lower Limit: "))
  xn = float(input("Upper Limit: "))
  h = (xn-x0)/n

  oddterms = 0
  for i in range (1, n, 2):
    oddterms = oddterms + 4*(f(x0+i*h))

  eventerms = 0
  for i in range(2, n, 2):
    eventerms = eventerms + 2*(f(x0+i*h))
```

```python
I = (1/3)* h * (f(x0) + oddterms + eventerms + f(xn))
print("Area: ", I)
```
