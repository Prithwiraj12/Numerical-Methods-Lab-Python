```python
def f(x): #function definition
  return x**2-2
```

```python
#Initial Setup
n = int(input("Enter Significant Figure n: "))
Es = 0.5*10**(2-n)
xl = 1
xu = 2
it = 1
Ea = 100
```

```python
while Ea >= Es:
  xr = (xl+xu)/2
  if f(xl)*f(xr) < 0:
    xu = xr
  elif f(xr)*f(xu) < 0:
    xl = xr

  cv = xr
  if it == 1:
    pv = cv
  elif it >= 2:
    Ea = abs((cv-pv)/cv)*100
    pv = cv
  print("Iteration ", it, " Ans: ", cv)
  it = it+1
```
