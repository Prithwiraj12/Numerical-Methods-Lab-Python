```python
def f(x):
  return 3*(x**2)+1
```

```python
a = float(input())
b = float(input())
N = int(input())
```

```python
import random
u = []
for i in range(N):
  u.append(random.random())

#print(u)
```

```python
x = []
for i in range(N):
  x.append(a+(b-a)*u[i])
#print(x)
```

```python
fx = []
for i in range(N):
  fx.append(f(x[i]))
#print(fx)
```

```python
sum = 0
for i in range(N):
  sum = sum + fx[i]

I = (b-a)*(sum/N)
print(I)
```
