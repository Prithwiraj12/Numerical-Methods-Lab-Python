```python
import numpy as np
```

**Vector Norm**

```python
x1 = [2, -3, 4]
x2 = np.array([2, -3, 4])
```

```python
#1-Norm
norm1D  = sum(abs(i) for i in x1)
norm1L = np.linalg.norm(x2, 1)

norm1D, norm1L
```

```python
#2-Norm
import math
norm2D = math.sqrt(sum(i**2 for i in x1))
norm2L = np.linalg.norm(x2, 2)
norm2D, norm2L
```

```python
#Inifnity-Norm
normiD = max(abs(i) for i in x1)
normiL = np.linalg.norm(x2, np.inf)
normiD, normiL
```

```python
#P-Norm
p = int(input())
normpD = (sum(abs(i)**p for i in x1))**(1/p)
normpL = np.linalg.norm(x2, p)
normpD, normpL
```

**Matrix Norm**

```python
A = [
    [1, -2],
    [3, 4]
]
nrows = len(A)
ncols = len(A[0])

col_sum  = []

for j in range(ncols):
  sum = 0
  for i in range(nrows):
    sum = sum + abs(A[i][j])
  col_sum.append(sum)

print("Matrix 1-Norm = ", max(col_sum))
```

```python
#Matrix Infinity Form
row_sum = []

for i in range(nrows):
  sum = 0
  for j in range(ncols):
    sum = sum+abs(A[i][j])
  row_sum.append(sum)
print("Matrix Infinity Norm = ", max(row_sum))
```

```python
#Matrix Frobenius Form
fro = 0

for row in A:
  for val in row:
    fro = fro + (val**2)
fro = math.sqrt(fro)

print("Matrix Forbenius Norm = ", fro)
```

```python
##Matrix Norm using numpy
print("1-Norm = " , np.linalg.norm(A, 1))
print("inf-Norm = ", np.linalg.norm(A, np.inf))
print("Frobenius-Norm = ", np.linalg.norm(A, 'fro'))
print("2-Norm = ", np.linalg.norm(A, 2))
```
