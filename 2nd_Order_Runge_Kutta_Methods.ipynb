```python
def fun(a2):
  a1 = 1-a2
  p1 = 0.5/a2
  q11 = 0.5/a2

  return a1, p1, q11
```

```python
def f(x, y):
  return x+y
```

```python
#Initial Setup
x = 0
y = 1
h = 0.05
xf = 0.1
```

```python
#Heun's Method
a2 = 1/2
a1, p1, q11 = fun(a2)
while x < xf:
  k1 = f(x, y)
  k2 = f(x+p1*h, y+q11*k1*h)
  y = y + (a1*k1+a2*k2)*h
  x = x+h
  print(x, y)
```
