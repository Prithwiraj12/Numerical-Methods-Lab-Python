```python
def fun(a2):
  a1 = 1-a2
  p1 = 0.5/a2
  q11 = 0.5/a2

  return a1, p1, q11
```

```python
def f(x, y):
  return x+y
```

```python
#Initial Setup
x = 0
y = 1
h = 0.05
xf = 0.1
```

```python
#Heun's Method
a2 = 1/2
a1, p1, q11 = fun(a2)
while x < xf:
  k1 = f(x, y)
  k2 = f(x+p1*h, y+q11*k1*h)
  y = y + (a1*k1+a2*k2)*h
  x = x+h
  print(x, y)
```

```python
#4th Order Runge-Kutta Method
#Initial Setup
x = 0
y = 1
h = 0.05
xf = 0.1
```

```python
while x < xf:
  k1 = f(x, y)
  k2 = f(x+0.5*h, y+0.5*k1*h)
  k3 = f(x+0.5*h, y+0.5*k2*h)
  k4 = f(x+h, y+k3*h)
  y = y+(1/6)*(k1+2*k2+2*k3+k4)*h
  x = x+h
  print(x, y)
```
