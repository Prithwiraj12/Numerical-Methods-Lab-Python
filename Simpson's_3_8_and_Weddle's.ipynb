```python
def f(x):
  return 0.2 + 25*x-200*(x**2)+675*(x**3)-900*(x**4)+400*(x**5)
```

```python
#Simpson's 3/8
n = 3
x0 = float(input())
x3 = float(input())
h = (x3-x0)/n
```

```python
I = (3/8) * h * (f(x0) + 3*f(x0+h) + 3*f(x0+2*h) + f(x3))
print("Area: ", I)
```

```python
#Simpson's 3/8 Multiple Segment
n = int(input()) # n must be multiple of 3
x0 = 0
xn = 0.8
h = (xn-x0)/n

term1 = 0
for i in range(1, n-1, 3):
  term1 = term1 + f(x0+i*h)


term2 = 0
for i in range(2, n, 3):
  term2 = term2 + f(x0+i*h)


term3 = 0
for i in range(3, n-2, 3):
  term3  = term3 + f(x0+i*h)


I = (3/8) * h * (f(x0) + 3*term1 + 3*term2 + 2*term3 + f(xn))
print("Area: ", I)
```

```python
#Weddle's Rule
n = 6 # n must be multiple of 6
x0 = 0
x6 = 0.8
h = (x6-x0)/6

I = (3/10)* h* (f(x0)+ 5*f(x0+h)+f(x0+2*h)+6*f(x0+3*h)+f(x0+4*h)+5*f(x0+5*h)+f(x6))
print("Area: ", I)
```

```python
#H/W: Multiple Segment Weddle's Rule
n = int(input())
x0 = 0
xn = 0.8
h = (xn-x0)/n
```

```python
eventerm = 0
for i in range(2, n, 2):
  if i%6 == 0:
    eventerm = eventerm + 2* f(x0 + i*h)
  else:
     eventerm = eventerm + f(x0 + i*h)
  print("f(x", i, ")")

sixterm = 0
for i in range(3, n, 6):
  sixterm = sixterm + f(x0 + i*h)
  print("f(x", i, ")")

fiveterm = 0
for i in range(1, n, 6):
  fiveterm = fiveterm + (f(x0+i*h)+f(x0+(i+4)*h))
  print("f(x", i, ")")

I = (3/10) * h * (f(x0) + eventerm + 5* fiveterm + 6* sixterm + f(xn))
print("Area: ", I)

```
