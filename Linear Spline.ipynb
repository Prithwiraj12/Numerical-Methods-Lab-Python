```python
n = int(input("Order of Polynomial: "))
x = []
fx = []
for i in range(n+1):
  x.append(float(input()))
for i in range(n+1):
  fx.append(float(input()))
```

```python
xp = float(input())
found = False
for i in range(n):
  if x[i] <= xp <= x[i+1]:
    fxp = fx[i] + ((fx[i+1] - fx[i]) / (x[i+1]-x[i]))*(xp - x[i]) #NDD
    fxp2 = ((xp-x[i+1])/(x[i]-x[i+1]))*fx[i] + ((xp-x[i])/(x[i+1]-x[i]))*fx[i+1] ##LM
    print(fxp, fxp2)
    found = True
    break
if not found:
  print("The point is outside the given ranges")
```
