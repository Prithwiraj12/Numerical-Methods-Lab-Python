```python
def lagrange(x, fx, xp):
  n = len(x)
  fxp = 0

  for i in range(n):
    L = 1
    for j in range(n):
      if i!= j:
        factor = (xp-x[j])/(x[i]-x[j])
        L = L * factor

    term = fx[i] * L
    fxp = fxp + term
  return fxp
```

```python
n = int(input("Order of polynomial: "))
x = []
fx = []
print("Enter x values: ")
for i in range(n+1):
  x.append(float(input()))

print("Enter fx values: ")
for i in range(n+1):
  fx.append(float(input()))
```

```python
xp = float(input("Enter Interpolation Point: "))
ans = lagrange(x, fx, xp)
print(ans)
```
