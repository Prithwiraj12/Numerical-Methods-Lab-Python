```python
#Gauss Jacobi
x1_old = 0.0
x2_old = 0.0
x3_old = 0.0
```

```python
it = 1
eax1 = 100
eax2 = 100
eax3 = 100
es = 0.05
```

```python
def sf(num):
  return float(f"{num:.{5}g}")
```

```python
while eax1 >= es or eax2 >= es or eax3 >= es:
  x1 = (0.1*x2_old+0.2*x3_old+7.85)/3.0
  x2 = (-1*0.1*x1_old+0.3*x3_old-19.3)/7.0
  x3 = (-1*0.3*x1_old+0.2*x2_old+71.4)/10.0
  eax1 = abs((x1 - x1_old)/x1)*100
  eax2 = abs((x2 - x2_old)/x2)*100
  eax3 = abs((x3 - x3_old)/x3)*100
  it = it+1
  print(sf(x1), sf(x2), sf(x3))
  x1_old = x1
  x2_old = x2
  x3_old = x3
```
