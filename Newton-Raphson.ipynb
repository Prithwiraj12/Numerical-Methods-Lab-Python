```python
#Initial Setup
xi = float(input())
n = int(input())
es = 0.5*10**(2-n)
ea = 100
it = 1
```

```python
def sf(num):
  return float(f"{num:.{n}g}")
```

```python
def f(x):
  return sf(x**2-3*x+1)
def fd(x):
  return sf(2*x - 3)
```

```python
while ea >= es:
  xi1 = sf(xi - (f(xi) / fd(xi)))
  ea = abs((xi1-xi)/xi1)*100
  print("Iteration No: ", it, "Root: ", xi1)
  xi = xi1
  it = it+1

```
